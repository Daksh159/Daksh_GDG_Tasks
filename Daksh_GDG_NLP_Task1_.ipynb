## Stemming

**What is Stemming?** <br>
**Ans:**
Stemming is a text normalization technique in Natural Language Processing (NLP) that reduces words to their root form by removing suffixes or prefixes. The resulting root word may not always be a valid word.

Stemming quickly chops off common word endings (-ing, -ed, -ly) to group related words together. It treats different forms of a word as the same to simplify text processing.

Example:

"Running" → "Run"

"Happiness" → "Happi"

"Studies" → "Studi"

Used when processing speed is more important than linguistic accuracy, such as in search engines or large-scale text analysis.

```python
# Create a word corpus from the sentences using word tokenization and stemming. Also, remember to avoid adding stopwords to your corpus
# Add your code here
corpus = []
for sentence in up_sen:
    words = nltk.word_tokenize(sentence)
    stemmed_words = [stemmer.stem(word) for word in words if word not in set(stopwords.words('english'))]
    corpus.append(' '.join(stemmed_words))
```

## Lemmatization

**What is Lemmatization?** <br>
**Ans:**Lemmatization is a text normalization technique in NLP that reduces words to their base or dictionary form (lemma) by analyzing the word's context, grammar, and meaning.


Lemmatization considers the correct part of speech and transforms the word into its proper, meaningful form.

Example:

"Running" → "Run"

"Better" → "Good"

"Studies" → "Study"

Used when accuracy and understanding of the text are important, such as in chatbots, virtual assistants, and machine translation.

```python
# Create a second word corpus from the sentences using word tokenization and lemmatization. Also, remember to avoid adding stopwords to your new corpus
# Add your code here
corp2 = []
for sentence in up_sen:
    words = nltk.word_tokenize(sentence)
    lemmatized_words = [lemmatizer.lemmatize(word) for word in words if word not in set(stopwords.words('english'))]
    corp2.append(' '.join(lemmatized_words))
```

**What is the difference between stemming and lemmatization? Is one better than the other? If yes, justify your answer. If no, determine the usecases where these can be used.** <br>
Ans:
**Stemming** and **lemmatization** are both techniques used in Natural Language Processing (NLP) to simplify words by reducing them to their base forms. However, they differ in how they achieve this.


**Stemming** is a rule-based process that removes prefixes or suffixes from words to reduce them to their root form. This method is fast but often crude, as it doesn't consider the meaning or grammatical context of words. For example, words like *"running"*, *"runner"*, and *"runs"* might all be reduced to *"run"*, but *"studies"* could become *"studi"*, which isn’t a valid word. Stemming is typically used in applications where speed is more important than precision, such as search engines or real-time data processing.

**Lemmatization**, on the other hand, is a more sophisticated approach. It reduces words to their base or dictionary form (*lemma*) by analyzing their grammatical context and meaning. For example, *"running"* becomes *"run"* and *"better"* becomes *"good"*. This process ensures that the result is always a valid word, making it more accurate than stemming. Lemmatization is preferred in tasks where understanding the meaning of words is crucial, such as in chatbots, machine translation, and text analysis.

Neither technique is strictly better than the other. **Stemming** is ideal when speed and simplicity are needed, while **lemmatization** is better suited for tasks that require accuracy and proper language understanding. The choice between them depends on the specific needs of the application.

```python
# Include all steps of removal of unnecessary characters, lowering, word tokenization, lemmatization and stopwords removal in one function
# Your final corpus should be a list of sentences, not a list of list of words (Why?)
# Add your code here
from nltk.tokenize import word_tokenize #Import the word_tokenize function

def preprocess_text(paragraph):
    stop_words = set(stopwords.words('english'))
    lemmatizer = WordNetLemmatizer()

    sentences = nltk.sent_tokenize(paragraph)

    clean_corp = []
    for sentence in sentences:
        # a. Remove unnecessary characters
        clean_sentence = re.sub(r'[^a-zA-Z0-9\s]', '', sentence)
        # b. Convert to lowercase
        clean_sentence = clean_sentence.lower()
        # c. Tokenize into words
        words = word_tokenize(clean_sentence)
        # d. Lemmatize and remove stopwords
        lemmatized_words = [
            lemmatizer.lemmatize(word) for word in words if word not in stop_words
        ]
        # e. Reconstruct the sentence and add to corpus
        clean_corp.append(' '.join(lemmatized_words))

    return clean_corp

# Preprocess the paragraph
corp = preprocess_text(par)
print("Cleaned Corpus:")
print(corp)

```

```python
# Now to play around with this:
def euclidean_distance(a_tf_idf_counter, b_tf_idf_counter) -> float:
    a_vec_sorted_keys = sorted(a_tf_idf_counter)
    b_vec_sorted_keys = sorted(b_tf_idf_counter)

    a_vec = [a_tf_idf_counter[key] for key in a_vec_sorted_keys]
    b_vec = [b_tf_idf_counter[key] for key in b_vec_sorted_keys]

    dist = 0
    for a_i, b_i in zip(a_vec, b_vec):
        dist += (b_i - a_i) ** 2
    return dist ** 0.5

# a (dot) b = abs(a) * abs(b) * cos(theta)
# cosine similarity is defined as a (dot) b / abs(a) * abs(b)
def cosine_similarity(a_tf_idf_counter, b_tf_idf_counter) -> float:
    a_vec_sorted_keys = sorted(a_tf_idf_counter)
    b_vec_sorted_keys = sorted(b_tf_idf_counter)

    a_vec = [a_tf_idf_counter[key] for key in a_vec_sorted_keys]
    b_vec = [b_tf_idf_counter[key] for key in b_vec_sorted_keys]

    dot_product = 0
    abs_a = 0
    abs_b = 0

    for a_i, b_i in zip(a_vec, b_vec):
        dot_product += a_i * b_i

        abs_a += a_i * a_i
        abs_b += b_i * b_i

    return dot_product / (abs_a ** 0.5 * abs_b ** 0.5)
```

# Compulsory Task


Research how the Bidirectional Encoder Representations from Transformers (BERT) Model creates Word Embeddings and how it maintains context and captures semantic meaning between words and sentences.

BERT creates word embeddings that understand the meaning of words in context, making it smarter than traditional models that treat all instances of a word the same. It represents words differently based on how they are used in a sentence. For example, in the sentence "The boat is near the bank," BERT knows that "bank" refers to a riverbank, but in "I deposited money in the bank," it understands that "bank" means a financial institution. This is possible because BERT processes the entire sentence both forward and backward, considering the relationship between all the words in the sentence.

BERT works by breaking sentences into smaller pieces, called tokens, which can handle even rare or unfamiliar words. It combines information about each token’s meaning, its position in the sentence, and the sentence structure to create dynamic representations of words. Using a mechanism called self-attention, BERT focuses on how each word connects to every other word, even if they are far apart in the sentence. This helps it capture not just the meaning of individual words but also the nuances of the entire sentence.

Because of this, BERT can handle complex language tasks like identifying sentiment, answering questions, or summarizing text with a deep understanding of the content. It’s especially useful when the meaning of a word depends heavily on the surrounding words, allowing it to understand sentences in a way that feels almost human.
